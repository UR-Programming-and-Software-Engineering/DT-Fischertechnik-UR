# Fischertechnik Lernfabrik with PLC documentation summary

This documentation is accompanied by SysML diagrams. Here a few words about what a SysML diagram is and what type of SysML diagrams you will encounter.

**SysML (Systems Modeling Language)** is a standardized modeling language used in systems engineering to specify, design, analyze, and verify complex systems. It extends UML (Unified Modeling Language) with additional diagrams tailored for engineering applications, such as hardware, software, and processes. SysML provides a visual way to represent system structure, behavior, requirements, and constraints. The language includes nine diagram types, categorized into structural diagrams that define system components and relationships, behavioral diagrams that describe system functionality and interactions, parametric diagrams that model mathematical constraints, and requirement diagrams that capture system specifications.

**The Block Definition Diagram (BDD)** is a structural diagram that defines the system architecture by representing components as blocks and their relationships. Its primary purpose is to define the hierarchy of system components, such as how a "Car" consists of an "Engine" and "Wheels," and to show inheritance, associations, and compositions between blocks. Key elements in a BDD include blocks, which represent system components like sensors or controllers, and relationships that illustrate how these components are connected.

**The Internal Block Diagram (IBD)** is another structural diagram that details how parts of a system interact internally. Its main purpose is to show internal connections between components, such as how a Battery supplies power to a Processor, and to define ports, interfaces, and data flows between parts. Key elements in an IBD include part properties, which are instances of blocks (e.g., :Battery, :Controller), connectors that represent wires, signals, or mechanical links between parts, and ports that serve as interaction points for electrical or data interfaces. For instance, an IBD for a Drone might illustrate how :Motor parts connect to a :FlightController through signal ports.

### Lernfabrik Ecosystem

In the fischertechnik Learning Factory 4.0 24V, three main components—the TXT 4.0 Controller, the IoT Gateway, and the PLC—work together to simulate an Industry 4.0 environment. Here's how they interact:

**TXT 4.0 Controller:** This microcontroller manages 9V components like environmental sensors, NFC readers, and USB cameras. It communicates using the MQTT protocol, sending data to the IoT Gateway.

**IoT Gateway (Raspberry Pi):** Acting as a bridge, the IoT Gateway receives MQTT messages from the TXT Controller and translates them into the OPC UA protocol for the PLC. It also facilitates communication between 9V and 24V components.

**PLC (Programmable Logic Controller)**: The PLC controls 24V components like conveyor belts and robotic arms. It receives data via OPC UA from the IoT Gateway and executes control logic accordingly.

Interaction Overview: Sensors collect data and send it to the TXT 4.0 Controller, which publishes this data using MQTT to the IoT Gateway. The IoT Gateway translates MQTT messages into OPC UA and forwards them to the PLC. The PLC processes the information and controls the 24V actuators accordingly.

This setup allows for seamless integration of various components, simulating real-world industrial automation scenarios.
